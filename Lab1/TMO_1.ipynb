Описание данных
Данный набор данных содержит информацию о характеристиках жилой недвижимости и соответствующих им ценах. Особенностью датасета является наличие как числового признака "Price" (цена в долларах), так и бинарной классификационной метки "Price_Label", которая разделяет объекты на две категории: дорогие и дешевые дома.

Датасет предназначен для решения задач как регрессии (предсказание точной цены дома), так и классификации (определение, относится ли дом к категории дорогих или дешевых).

Имеет небольшой размер (600 записей), что позволяет быстро выполнять вычисления
Не содержит пропусков
Включает как числовые, так и категориальные признаки

# Визуальное исследование датасета

```python
data.hist(figsize=(12, 10), bins=20, edgecolor='black')
plt.suptitle('Гистограммы распределения числовых признаков', fontsize=14)
plt.show()
```

```python
# Зависимость цены от площади
plt.figure(figsize=(10, 6))
plt.scatter(data['Size'], data['Price'], alpha=0.6)
plt.xlabel('Площадь')
plt.ylabel('Цена')
plt.title('Зависимость цены от размера дома')
plt.grid(True, alpha=0.3)
plt.show()
```

# Простая матрица корреляций

```python
import seaborn as sns
# Выбираем только числовые колонки
numeric_cols = ['Size', 'Bedrooms', 'Price']
corr_matrix = data[numeric_cols].corr()

print("Матрица корреляции:")
print(corr_matrix)
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Матрица корреляции признаков')
plt.show()
```
